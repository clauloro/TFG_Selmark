```python
import pandas as pd
from pathlib import Path

RUTA_RAW = Path("..") / "data" / "raw"

archivos = [
    "dim_cliente.csv",
    "fact_lineas_pedido.csv",
    "mosaic.csv",
    "ventas_minoristas.csv",
]

print("VERIFICACION DE FILAS POR ARCHIVO\n")
print(f"{'Archivo':<35} {'Filas':>12}  {'Tamaño MB':>10}")
print("-" * 65)

for archivo in archivos:
    ruta = RUTA_RAW / archivo
    if ruta.exists():
        tamanio_mb = ruta.stat().st_size / (1024 * 1024)
        try:
            df = pd.read_csv(ruta, sep=",", low_memory=False)
            n_filas = len(df)
        except Exception:
            try:
                df = pd.read_csv(ruta, sep=";", low_memory=False)
                n_filas = len(df)
            except Exception as e:
                n_filas = "ERROR"
        print(f"{archivo:<35} {n_filas:>12,}  {tamanio_mb:>10.2f}")
    else:
        print(f"{archivo:<35} NO ENCONTRADO")
```
